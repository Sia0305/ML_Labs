# Dataset: `Iris.csv`
# 1. Import Data: Load the Iris dataset using `sklearn.datasets` or CSV.

```python
from sklearn.datasets import load_iris
import pandas as pd

# load dataset
iris = load_iris()

#create dataFrame
df = pd.DataFrame(iris.data , columns=iris.feature_names)

#Add target columns
df['species'] = iris.target

#Display
df.head()
```

# 2. Exploration: Display the feature names and target classes.

```python
print(iris.feature_names)
print(iris.target_names)
```

# 3. Split Data: Split the data into training and testing sets.

```python
from sklearn.model_selection import train_test_split

x = df.drop('species',axis = 1)
y = df['species']
```

```python
X_train ,X_test , y_train , y_test = train_test_split(x,y,test_size = 0.2 , random_state=42)
```

# 4. Train Model: Initialize a Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(criterion = 'gini')
```

# 5. Fit: Train the classifier on the training data.

```python
model.fit(X_train ,y_train)
```

# 6. Predict: Make predictions on the test set.

```python
y_pred = model.predict(X_test)
y_pred
```

# 7. Accuracy: Calculate the accuracy score.

```python
from sklearn.metrics import accuracy_score

acc = accuracy_score(y_pred,y_test)
print('model accuracy',acc)
```

# 8. Save Model Using `Joblib` Library

 joblib is a Python library used primarily to save (serialize) and load (deserialize) trained models so you do not have to retrain them from scratch. It serves as a faster, more memory-efficient alternative to Python's built-in pickle module because it is specifically optimized to handle large datasets and internal NumPy arrays. Beyond model saving, joblib is also heavily used for disk-based caching and parallel computing.

```python
import joblib

joblib.dump(model, "iris_decision_tree.pkl")
```
